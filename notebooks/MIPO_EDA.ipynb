```python
import json
import subprocess
import sys
from pathlib import Path

from IPython.display import SVG, display

root = Path.cwd()
while not (root / "scripts/eda_mipo.py").exists():
    if root.parent == root:
        raise RuntimeError("Open notebook inside P-063 checkout")
    root = root.parent
command = [sys.executable, str(root / "scripts/eda_mipo.py")]
if not (root / "data/mipo/sample.jsonl").exists():
    command.append("--download")  # pinned source and SHA-256 checks; requires network
result = subprocess.run(command, cwd=root, check=True, capture_output=True, text=True)
print(result.stdout)
profile = json.loads((root / "notebooks/mipo/profile.json").read_text())
```
